```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# read all the words
words = open('names.txt', 'r').read().splitlines()

words[:8]
```

```python
len(words)
```

```python
# build the vocabulary of characters and mappings to/from integers

chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
print(itos)
```

```python
# build the dataset

block_size = 3
X, Y = [], []
#for w in words[:5]:
for w in words:
    #print(w)
    context = [0] * block_size
    for ch in w + '.':
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        #print(''.join(itos[i] for i in context), '----->', itos[ix])
        context = context[1:] + [ix]
        
X = torch.tensor(X)
Y = torch.tensor(Y)
```

So X contains an array of arrays. The outer array has an element for each pattern to match. Each inner array is 3 elements, which contain the indices of the 3 characters in the match pattern.
